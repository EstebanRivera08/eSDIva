# 3. 3D visualization of the brain

Let's plot a 3D visualization with of the selected atlas.

To do so, we create a function that converts the atlas'mesh to a pyvista object.

````python
def meshio_to_pv(
    mesh_io,
    *,
    translation=None,
    rotate_x=0,
    rotate_y=0,
    rotate_z=0,
    order = "xyz",
    flip_z=False,
    scaling_factor=1e-3,  # μm → mm conversion 
):  
    """Convert meshio.Mesh → pyvista.PolyData with transformations."""
    verts = mesh_io.points.copy()  # Work on a copy to preserve original data
    tris = mesh_io.cells_dict.get("triangle", [])

    if len(tris) == 0:
        return None
    
    # 1. Convert to millimeters first
    verts *= scaling_factor
    
    # Create temporary PolyData for transformations
    polydata = pv.PolyData(
        verts,
        np.hstack([np.full((tris.shape[0], 1), 3, dtype=int), tris]).ravel()
    )
    
    # 2. Rotations (order matters: X → Y → Z)
    if order == "xyz":
        polydata.rotate_x(rotate_x, inplace=True)
        polydata.rotate_y(rotate_y, inplace=True)
        polydata.rotate_z(rotate_z, inplace=True)
    elif order == "zyx":
        polydata.rotate_z(rotate_z, inplace=True)
        polydata.rotate_y(rotate_y, inplace=True)
        polydata.rotate_x(rotate_x, inplace=True)
    elif order == "xzy":
        polydata.rotate_x(rotate_x, inplace=True)
        polydata.rotate_z(rotate_z, inplace=True)
        polydata.rotate_y(rotate_y, inplace=True)
    elif order == "yxz":
        polydata.rotate_y(rotate_y, inplace=True)
        polydata.rotate_x(rotate_x, inplace=True)
        polydata.rotate_z(rotate_z, inplace=True)
    elif order == "yzx":
        polydata.rotate_y(rotate_y, inplace=True)
        polydata.rotate_z(rotate_z, inplace=True)
        polydata.rotate_x(rotate_x, inplace=True)
    elif order == "zxy":
        polydata.rotate_z(rotate_z, inplace=True)
        polydata.rotate_x(rotate_x, inplace=True)
        polydata.rotate_y(rotate_y, inplace=True)
    else:
        raise ValueError(f"Invalid order: {order}. Use 'xyz', 'zyx', 'xzy', 'yxz', 'yzx', or 'zxy'.")

    # # 3. re-center
    if translation is not None:
        polydata.translate(translation, inplace=True)

    return polydata
````

## 4.1. Functions

### 4.1.1. Import matlab structures

````python
from scipy.io import loadmat

def import_matlab(filepath):
    mat_file_struct = loadmat(filepath)
    setup_simu = mat_file_struct['setup_simu']
    field_struct = mat_file_struct['field_source']
    medium = mat_file_struct['medium']
    return setup_simu, field_struct, medium
````

### 4.1.2. Create transducer mesh

````python
import numpy as np
import pyvista as pv

def create_transducer_surface(data, scaling_factor=1000):
    """
    Create transducer surface mesh from data array.
    
    Parameters:
    - data: 2D numpy array with shape (22, M) where M is number of elements
    - scaling_factor: Scale factor for vertex positions (default 1000 for m→mm)
    
    Returns:
    - pyvista.PolyData mesh with apodization colors
    """
    M = data.shape[1]
    
    # Initialize arrays
    vertices = np.zeros((4 * M, 3))  # 4 vertices per element
    faces = np.zeros((M, 5), dtype=int)  # PyVista quads need 5 elements per face [4, v1, v2, v3, v4]
    colors = np.zeros(M)

    for i in range(M):
        # Get vertex indices for current element
        v_idx = i * 4
        
        # Extract and scale vertices (MATLAB 1-based → Python 0-based)
        vertices[v_idx] = data[10:13, i] * scaling_factor    # BL (11-13 in MATLAB) Bottom Left
        vertices[v_idx+1] = data[13:16, i] * scaling_factor  # BR (14-16)  # Bottom Right
        vertices[v_idx+2] = data[16:19, i] * scaling_factor  # TR (17-19) # Top Right
        vertices[v_idx+3] = data[19:22, i] * scaling_factor  # TL (20-22) # Top Left

        # Create face (quad) - PyVista format: [4, v1, v2, v3, v4]
        faces[i] = [4, v_idx, v_idx+1, v_idx+2, v_idx+3]
        
        # Store apodization value
        colors[i] = data[4, i]  # MATLAB data(5,i) → data[4,i]

    # Create PyVista mesh
    mesh = pv.PolyData(vertices, faces.ravel())
    mesh["Apodization"] = colors  # Add scalar array for coloring
    
    return mesh
````

### 4.1.3. Get 3D pressure field

````python
def mirror_yaxis_python(amplitude_half_field):
    """
    Mirror a 3D field along the y-axis and combine the original and mirrored fields.

    Parameters:
    -----------
    amplitude_half_field : np.ndarray
        A 3D numpy array representing the half field to be mirrored.

    Returns:
    --------
    amplitude_field_3D : np.ndarray
        A 3D numpy array with the field mirrored along the y-axis.
    """
    # Get the dimensions of the input field
    dim1, dim2, dim3 = amplitude_half_field.shape

    # Create a new tensor with double the size in the y-dimension
    amplitude_field_3D = np.zeros((dim1 * 2 - 1, dim2, dim3))

    # Mirror along the y-axis
    tensor_y_mirrored = np.flip(amplitude_half_field, axis=0)

    # Combine the original tensor with the mirrored tensor
    # Place the y-mirrored tensor in the first part
    amplitude_field_3D[:dim1 - 1, :, :] = tensor_y_mirrored[1:, :, :]

    # Place the original tensor in the second part
    amplitude_field_3D[dim1 - 1:, :, :] = amplitude_half_field

    amplitude_field_3D = np.transpose(amplitude_field_3D, (1, 0, 2))
    return amplitude_field_3D
````

````python

def compute_Pr_Field(field_struct):
    """
    Import a MATLAB file and extract the 'field_source' structure.

    Parameters:
    -----------
    filepath : str
        Path to the MATLAB file.

    Returns:
    --------
    x : np.ndarray
        X-coordinates of the field.
    y : np.ndarray
        Y-coordinates of the field.
    z : np.ndarray
        Z-coordinates of the field.
    scaled_field : np.ndarray
        Scaled field data.
    """

    # If 'field_source' is a structured array, navigate through its fields
    if isinstance(field_struct, np.ndarray) and field_struct.dtype.names:
        # Extract the 3D matrix (adjust the field name as needed)
        scaled_half_field = field_struct['scaled_half_field'][0][0]
        scaled_field = mirror_yaxis_python(scaled_half_field)**2  # Square the field to get the PII
        x = field_struct['x_vec'][0][0].squeeze()*1e3  # Convert to mm
        y = field_struct['y_vec'][0][0].squeeze()*1e3  # Convert to mm
        z = field_struct['z_vec'][0][0].squeeze()*1e3  # Convert to mm

    return x, y, z, scaled_field
````

### 4.1.4. Plot 3D pressure distribution

````python
def create_pressure_volume(field_struct):
    """
    Create a 3D volume mesh from the pressure field data.

    Parameters:
    -----------
    field_struct : 
        Structured array containing the pressure field data.
        It should have fields 'x_vec', 'y_vec', 'z_vec', and 'scaled_half_field'.
    Returns:
    --------
    pv.PolyData
        PyVista PolyData object representing the pressure volume.
    """
    x,  y, z, pressure_field = compute_Pr_Field(field_struct)
    # If x, y, z are 1D (common case)
    dx = x[1] - x[0]
    dy = y[1] - y[0]
    dz = z[1] - z[0]

    nx, ny, nz = pressure_field.shape

    # Create the 3D UniformGrid
    pressure_vol = pv.ImageData(dimensions=(nx, ny, nz),
                        spacing=(dx, dy, dz),
                        origin=(x.min(), y.min(), z.min()))

    # Attach pressure data to the grid
    pressure_vol.point_data["Pressure"] = pressure_field.ravel(order="F")  # VERY important: Fortran order
    return pressure_vol
````
